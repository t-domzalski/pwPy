````python
def encrypt_decrypt(number_to_process):

  encryption_key = dict(zip(list(range(10)), list(range(9, -1, -1))))
  decryption_key = dict(zip(encryption_key.values(), encryption_key.keys()))

  encrypted = ""
  for e in str(number_to_process):
    encrypted += str(encryption_key[int(e)])
  print(encrypted)

  decrypted = ""
  for e in encrypted:
    decrypted += str(decryption_key[int(e)])
  print(decrypted)

  return int(decrypted)
````

````python
def is_odd(x):
  if x % 2 == 0:
    return False

  else:
    return True

def is_list_odd(n_list):
  result = []

  for el in n_list:
    result.append(is_odd(el))

  return result
````

````python
l = [1, 2, 5, 9, 7, 0.3]

is_list_odd(l)
````
