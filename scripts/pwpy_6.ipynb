````python
lista = [10, 20, 30, 20, 10, 50, 60, 40, 80, 50, 40]
````

````python
unikalne = []
indeksy = []
````

````python
for index, element in enumerate(lista):

  #licznik ile razy pojawil sie dany element
  licznik = 0

  for el in range(len(lista)):

    if lista[el] == element:
      licznik += 1

  if licznik == 1:

    unikalne.append(element)
    indeksy.append(index)

````

````python
plansza = "---------"
gracz = "x"

while True:

  print(plansza[:3], plansza[3:6], plansza[6:9], sep="\n")

  decyzja = input(f"Gdzie postawic '{gracz}': ")

  if not (decyzja.isdigit() and 1 <= int(decyzja) <= 9 and plansza[int(decyzja)-1] == "-"):
    print("zleeeeee, podaj numer pustego pola (od 1 do 9)")
    continue


  index = int(decyzja) - 1

  plansza = plansza[:index] + gracz + plansza[index + 1:]

  warianty = [
      plansza[:3],
      plansza[3:6],
      plansza[6:9],
      plansza[:7:3],
      plansza[1:8:3],
      plansza[2:9:3],
      plansza[:9:4],
      plansza[2:7:2]
  ]

  if "xxx" in warianty or "ooo" in warianty:

      print(plansza[:3], plansza[3:6], plansza[6:9], sep="\n")

      print(f"Wygrywa '{gracz}'")

      break

  if "-" not in plansza:

      print(plansza[:3], plansza[3:6], plansza[6:9], sep="\n")

      print("Plansza pełna")

      break

  if gracz == "x":
      gracz = "o"

  else:
      gracz = "x"

````

List comprehension



[wyrażenie FOR element IN iterowalna_struktura]

````python
a = [1,2,3,4,5]
res = []

for element in a:
  res.append(element * 2)

print(res)
````

````python
res = [element * 2 for element in a]
print(res)
````

````python
l1 = [v**2 for v in range(10)]
l1
````

````python
even = []

for n in a:
  if n % 2 == 0:
    even.append(n)

even
````

[wyrażenie FOR element IN iterowalna_struktura IF condition]

````python
even = [n for n in a if n % 2 == 0]
even
````

````python
text = "ala ma kota albo nie ma kota"
text_po = [character for character in text]
text_po
````

````python
text = "ala ma kota albo nie ma kota nikogo to nie obchodzi"
letters = [character.upper() for character in text]
letters
````

````python
p = [1, 3, 6, 9]
q = ['a', 'b', 'c', 'd']

odp = [p[element] * q[element] for element in range(len(p))]
odp
# ["a", "bbb", "cccccc", "ddddddddd"]
````

````python
l2 = [n for n in range(20) if n % 2 == 0 if n % 3 == 0]
l2
````

````python
nums = [1, 2, 3, 4, 5]
res = [n if n % 2 == 0 else -n for n in nums]

res
````

````python
l3 = [n * 10 for n in [n ** 2 for n in range(10)]]
l3
````
