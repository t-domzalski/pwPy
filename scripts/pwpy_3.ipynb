Napisz kod, który sprawdzi wartosc zmiennej number i wypisze odpowiedni komunikat:
- "Wieksza niz lub równa 5!" jezeli number jest wieksze niz lub równe 5
- "Pomiedzy 2, a 5!" jezeli number nalezy do zakresu od 2 do 5 (przedział otwarty)
- "Mniejsza niz lub równa 2!" jezeli number jest mniejsze niz lub równe 2

````python
number = input("podaj dowolna liczbe")
if int(number) >= 5:
  print("Wieksza niz lub równa 5!")
elif 2 <= int(number) >= 5:
  print("Pomiedzy 2, a 5!")
else:
  print("Mniejsza niz lub równa 2!")
````
