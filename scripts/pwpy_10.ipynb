```python
# Zadanko
# Stworz funkcje replace_many, ktora podmienic wszystkie
# wystapienia slow przkazanych jako lista
# funkcja powinna przyjmowac jako argumenty tekst, lista_starych, lista_nowych
# zwraca nam nowy tekst z dokonanymi podmianami
# "Ala ma kota", lista stare = ["Ala", "kota"] i nowe = ["Edyta", "psa"]
# Edyta ma psa jako nowe
```

```python
def replace_many(tekst: str, lista_starych: list[str], lista_nowych: list[str]) -> str:
  """
  Przyklad, krotkiego docsa
  Funkcja podmieniajaca slowa w danym stringu na nowe

  Args:

    tekst (str): *nie trzeba tu dodawac str,
                bo widac w def parametrow* tekst w stringu
    lista_starych (list[str]) * to samo: lista słow w tekscie oryginalnym, ktore chcemy podmienic
    lista_nowych (list[str]) * to samo: lista słow, ktore chcemy umiescic w miejscu slow w 'lista_starych'

  Returns:
    str * to samo: nowy tekst po dokonaniu podmian

  """

  # Kod jest czesto komentowany, komentarze bywaja lepsze i gorsze.
  # To co jest najbardziej uzyteczne, to pisanie "dlaczego" w komentarzach
  # a nie co dana linijka kodu robi, czesto po kodzie widać co on robi.
  # Jesli obawiasz się, ze nie bd pamietac co robi za jakis czas to mozna to napisac, oczywiscie.
  # Niemniej zawsze warto napisać dlaczego, np:
  # Co: petla iteruje po elementach starej listy (widac po kodzie)
  # Wersja 'dlaczego': Iterujemy po petli wykorzystujac indexy, bo potrzebujemy
  # prezyzyjnie polaczyc pary, stare - nowe slowo,
  # uzywajac ich lokalizacji w odpowiadajacych im sekwencjach
  # Komentarze lepiej kiedy tłumacza intencje programisty, anizeli skladnie kodu
  # to podejscie wymaga wiecej pracy przy dokumentacji, a

  for i in range(len(lista_starych)):
    tekst = tekst.replace(lista_starych[i], lista_nowych[i])

  return tekst
```

```python
print(replace_many("Ala ma kota", ["Ala", "kota"], ["Edyta", "psa"]))
```
