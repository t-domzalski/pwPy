Zadanko 2

```python
rekordy = [
    {"imie": "Adam", "nazwisko": "Nowak", "wiek": 35},
    {"imie": "Anna", "nazwisko": "Nowak", "wiek": 29},
    {"imie": "Artur", "nazwisko": "Nowakowski", "wiek": 39}
]
```

```python
# Stworzyc nowa liste slownikow,
# tak aby slowniki zawieraly tylko klucze imie i nazwisko
```

```python
new_rekordy = []
```

```python
for osoba in rekordy:
  new_rekordy.append({osoba['imie'], osoba['nazwisko']})

print(new_rekordy)
```

```python
print(type(new_rekordy[0]))
```

```python
new_rekordy2 = []

for osoba in rekordy:

  new_dict = {
      "imie": osoba["imie"],
      "nazwisko": osoba["nazwisko"]
  }

  new_rekordy2.append(new_dict)

new_rekordy2
```
