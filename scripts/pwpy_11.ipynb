```python
# Zdefiniuj klase Samochod z konstruktorem, ktorej instancje beda posiadaly
# atrybuty przekazywane jako argumenty
# marka, kolor, stan_licznika z domyslna wartoscia 0
# chcemy instancje w postaci bialy merol prosto z salonu (licznik = 0)
# druga instancje czarna bejca z 150000 na liczniku
```

```python
class Samochod:
  def __init__(self, marka, kolor, stan_licznika = 0):
    self.marka = marka
    self.kolor = kolor
    self.licznik = stan_licznika
```

```python
merc = Samochod("merc", "bialy")
bejca = Samochod("bejca", "czarna", stan_licznika=150000)
```

```python
merc.licznik
```

```python
merc.licznik += 50
```

```python
merc.licznik
```

```python
trasa = [10, 156, 106, 869, 6]
```

```python
for fragment in trasa:
  bejca.licznik += fragment
```

```python
bejca.licznik += sum(trasa)
```
