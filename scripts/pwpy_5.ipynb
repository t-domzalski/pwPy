````python
numbers = [3, 5, 11, 12, 1]

# cos sie dzieje przed petla

for number in numbers: # daje 11
  print(number) # printuje nam 11

  if number > 10: # sprawdza czy 11 > 10

    print(number) # printujemy 11

    break # zamykamy petle

#cos po petli
````

````python
numbers = [[1, 3, 9], [3, 2], [4, 2, 1, 3]]

for group in numbers:

  if len(group) != 2:
    continue

  elif sum(group) <= 5:
    print(group, "grupa z suma nie wieksza niz 5")
````

# Zadanie 2

Napisz program, który sprawdzi, czy dwie listy mają przynajmniej jeden element wspólny. Wypisz True lub False.

````python
list_a = [1, 23, 94, "kot", "pies", [2,3]]
list_b = [54, "kot", 18]
````

````python
wynik = False

for element in list_a:

  if element in list_b:

    wynik = True

print(wynik)
````

````python
i = 0
wynik = False

while i < len(list_a):

  if list_a[i] in list_b:

    wynik = True

  i += 1

print(wynik)
````

````python
list_a = [1, 23, 94, "kot", "pies", [2,3]]
list_b = [54, "kot", 18]

for element in list_a:

  if list_b.count(element) == 0:
    continue

  else:
    print(element, "True")
````

````python
list_a = [1, 23, 94, "kot", "pies", [2,3]]
list_b = [54, "kot", [2, 3]]
````

````python
wynik = False

for element in list_a:

  if element in list_b:

    wynik = True

print(wynik)
````

````python
list_a = [1, 23, 94, "kot", "pies", [2,3]]
list_b = [54, "kot", [2,3]]

for element in list_a:

  if list_b.count(element) == 1:
    break

print(element, "True")
````

# Zadanie 3

Stwórz nową listę, w której imiona będą miały dopisane " TAK" lub " NIE" w zależności od tego czy znajdują się na liście obecności

````python
wszyscy = ["Anna", "Jurek", "Beata", "Jarek", "Patrycja", "Arkadiusz"]
obecni = ["Jurek", "Patrycja", "Beata"]
````

````python
raport = []

for imie in wszyscy:

  if imie in obecni:

    raport.append([imie, "TAK"])

  else:
    raport.append([imie, "NIE"])
````
