```python
# Sumowac kwadraty kilku liczb w sekwencji, l = [1, 2, 3]
```

```python
def fun(l):

  sum_sq = 0

  for i in l:
    sum_sq += i ** 2


  return sum_sq
```

```python
def fib(n):

  print("fib", n)

  if n <= 1:
    return n

  return fib(n - 1) + fib(n - 2)
```

```python
fib(4)
```

```python
# args i kwargs
```

```python
def func(*args):
  print(args[0][1])
  for i in args:
    print(i)
```

```python
l_a = [1, 2, 3]
```

```python
func(l_a)
```

```python
func("hey", "you")
```

```python
def func(*args, **kwargs):

  print(args[0][2])
  for i in args:
    print(i)

  for key, value in kwargs.items():
    print(f"{key} == {value}")
```

```python
func("mama", "tata", kon = "rafal", swinka = "pepa")
```

```python
# Kalkulator BMI
# bmi = waga / (wzrost ** 2)
```

```python
def calc_bmi(waga, wzrost):

  bmi = waga / ((wzrost/100)**2)

  return bmi
```

```python
calc_bmi(100, 190)
```

```python
# jesli output bedzie ustawiony na kategorie, niedowaga 18.5 >= bmi
# norma 18.5 < bmi < 25
# nadwaga 25 <= bmi < 30
# otylosc bmi >= 30
```

```python
def calc_bmi(waga, wzrost, output = None):

  bmi = waga / ((wzrost/100) ** 2)

  if output is None:
    return round(bmi, 2)

  elif output is "kategorie":

    if bmi <= 18.5:
      return "Niedowaga"

    if 18.5 < bmi < 25:
      return "Norma"

    if 25 <= bmi < 30:
      return "Nadwaga"

    if bmi >= 30:
      return "Otylosc"

```

```python
calc_bmi(70, 180)
```

```python
calc_bmi(70, 180, output = "kategorie")
```
