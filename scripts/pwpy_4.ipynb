operator `in` można wykorzystać do weryfikacji `single object IN sequence`

```python
lista1 = ['apple', 'banana', "fruit3"]
target = ["apple", "fruit3", "jajko"]

for e in target:
  if e in lista1:
    print(e)

  else:
    print(e, "not on the list")
```

Zadanie 2

1. Stwórz listę nagród składającą się z 3 elementów.  
2. Wyświetl całą listę nagród.  
3. Zapytaj użytkownika, czy chce wymienić ostatnią nagrodę na batona czekoladowego.  
4. Jeśli użytkownik odpowie „tak”, dokonaj wymiany ostatniej nagrody; jeśli „nie”, pozostaw listę bez zmian.  
5. Na końcu wyświetl efekt końcowy – aktualną listę nagród.

```python
nagrody = ["książka", "czekolada", "puchar"]
wymiana = input("Wymieniasz?: ")

if wymiana == "Tak":
    nagrody[-1] = "baton czekoladowy"
    print(f'Wybrane nagrody{nagrody}')
else:
    print(nagrody)
```

Hint: operator % do sprawdzania reszty z dzielenia

```python
b = 1

while b <= 100:
  if b % 2 != 0 and b % 3 != 0 and b % 4 != 0:
    print(b)
  b += 1
```
